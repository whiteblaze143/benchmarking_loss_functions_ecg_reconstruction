# Paper Metrics Suite

This notebook generates the final LaTeX tables and figures for the publication.
It ingests `evaluation_results.json` and formats it into publication-ready artifacts.


```python
import pandas as pd
import json

RESULTS_PATH = "../logs/evaluation_results.json"
MODELS_ORDER = ["mason_cnn", "mason_ricker", "cnvae_ecg", "cnvae_ricker"]
MODEL_NAMES = {
    "mason_cnn": "Mason-CNN (Control)",
    "mason_ricker": "Mason-NIWT (Hybrid)",
    "cnvae_ecg": "cNVAE-ECG (SOTA)",
    "cnvae_ricker": "cNVAE-NIWT (Ours)"
}

try:
    with open(RESULTS_PATH, 'r') as f:
        raw = json.load(f)
    
    rows = []
    for m in MODELS_ORDER:
        if m in raw and "seed_42" in raw[m]:
            r = raw[m]["seed_42"]
            rows.append({
                "Model": MODEL_NAMES[m],
                "MAE (mV)": f"{r['test_mae']:.4f}",
                "MSE": f"{r['test_mse']:.4f}",
                "Pearson": f"{r['test_corr']:.3f}",
                "AUROC (Clinical)": f"{r['auroc']:.3f}"
            })
            
    df = pd.DataFrame(rows)
    print("Generated Table:")
    display(df)
    
    # To Latex
    print("\nLaTeX Code:")
    print(df.to_latex(index=False, caption="Comparative Performance on PTB-XL Test Set", label="tab:main_results"))
    
except FileNotFoundError:
    print("Results file not found. Run evaluation script first.")
```
