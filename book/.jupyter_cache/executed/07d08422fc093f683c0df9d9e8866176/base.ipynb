```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class DoubleConv1D(nn.Module):
    """(Convolution -> BatchNorm -> LeakyReLU) * 2"""
    def __init__(self, in_channels, out_channels):
        super().__init__()
        self.double_conv = nn.Sequential(
            nn.Conv1d(in_channels, out_channels, kernel_size=15, padding=7, bias=False),
            nn.BatchNorm1d(out_channels),
            nn.LeakyReLU(0.2, inplace=True),
            nn.Conv1d(out_channels, out_channels, kernel_size=15, padding=7, bias=False),
            nn.BatchNorm1d(out_channels),
            nn.LeakyReLU(0.2, inplace=True)
        )

    def forward(self, x):
        return self.double_conv(x)

class UNet1DECG(nn.Module):
    """
    1D U-Net for 12-Lead ECG Reconstruction.
    Input Channels: 3 (Reduced Leads)
    Output Channels: 12 (Full Clinical Array)
    """
    def __init__(self, in_channels=3, out_channels=12, base_filters=64):
        super().__init__()
        # Encoder (Contracting Path)
        self.inc = DoubleConv1D(in_channels, base_filters)
        self.down1 = nn.Sequential(nn.MaxPool1d(2), DoubleConv1D(base_filters, base_filters * 2))
        self.down2 = nn.Sequential(nn.MaxPool1d(2), DoubleConv1D(base_filters * 2, base_filters * 4))
        self.down3 = nn.Sequential(nn.MaxPool1d(2), DoubleConv1D(base_filters * 4, base_filters * 8))

        # Bottleneck
        self.bottleneck = DoubleConv1D(base_filters * 8, base_filters * 8)

        # Decoder (Expanding Path)
        self.up1 = nn.ConvTranspose1d(base_filters * 8, base_filters * 4, kernel_size=2, stride=2)
        self.conv_up1 = DoubleConv1D(base_filters * 8, base_filters * 4)

        self.up2 = nn.ConvTranspose1d(base_filters * 4, base_filters * 2, kernel_size=2, stride=2)
        self.conv_up2 = DoubleConv1D(base_filters * 4, base_filters * 2)

        self.up3 = nn.ConvTranspose1d(base_filters * 2, base_filters, kernel_size=2, stride=2)
        self.conv_up3 = DoubleConv1D(base_filters * 2, base_filters)

        # Output Lead Projection
        self.outc = nn.Conv1d(base_filters, out_channels, kernel_size=1)

    def forward(self, x):
        # Encoder
        x1 = self.inc(x)
        x2 = self.down1(x1)
        x3 = self.down2(x2)
        x4 = self.down3(x3)

        # Bottleneck
        b = self.bottleneck(x4)

        # Decoder with Skip Connections
        x = self.up1(b)
        x = torch.cat([x, x3], dim=1)
        x = self.conv_up1(x)

        x = self.up2(x)
        x = torch.cat([x, x2], dim=1)
        x = self.conv_up2(x)

        x = self.up3(x)
        x = torch.cat([x, x1], dim=1)
        x = self.conv_up3(x)

        logits = self.outc(x)
        return logits

# Instantiate and verify model parameters
model = UNet1DECG(in_channels=3, out_channels=12, base_filters=64)
dummy_input = torch.randn(4, 3, 1000)
dummy_output = model(dummy_input)

num_params = sum(p.numel() for p in model.parameters())
print(f"1D U-Net Model Initialized!")
print(f"Input Shape: {dummy_input.shape} -> Output Shape: {dummy_output.shape}")
print(f"Total Trainable Parameters: {num_params:,}")
```
