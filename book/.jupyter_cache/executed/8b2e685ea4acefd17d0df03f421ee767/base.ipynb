```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import scipy.signal as signal

class SyntheticECGDataset(Dataset):
    """
    Simulates a 12-lead ECG dataset with cardiac P-QRS-T components.
    """
    def __init__(self, num_samples=200, seq_len=1000, sampling_rate=500):
        self.num_samples = num_samples
        self.seq_len = seq_len
        self.sampling_rate = sampling_rate
        
        t = np.linspace(0, seq_len / sampling_rate, seq_len)
        data = []
        for i in range(num_samples):
            hr = np.random.uniform(60, 90) # Heart rate
            freq = hr / 60.0
            
            # Base cardiac waveform components
            qrs = np.sin(2 * np.pi * freq * t) ** 21 # Sharp QRS spike (High freq)
            p_wave = 0.2 * np.sin(2 * np.pi * freq * t - 0.5) ** 5 # P wave (Low freq)
            t_wave = 0.35 * np.sin(2 * np.pi * freq * t + 0.8) ** 3 # T wave (Low freq)
            
            # Baseline wander (Low frequency drift ~0.5 Hz)
            baseline_wander = 0.3 * np.sin(2 * np.pi * 0.5 * t)
            
            leads = []
            for lead_idx in range(12):
                phase_shift = lead_idx * 0.15
                amp_scale = 0.5 + 0.5 * np.cos(lead_idx * 0.5)
                # High frequency noise
                noise = np.random.normal(0, 0.05, seq_len)
                
                lead_signal = amp_scale * (
                    np.sin(2*np.pi*freq*t + phase_shift)**21 + 
                    0.2*np.sin(2*np.pi*freq*t - 0.5 + phase_shift)**5 + 
                    0.35*np.sin(2*np.pi*freq*t + 0.8 + phase_shift)**3
                ) + noise + baseline_wander
                leads.append(lead_signal)
            
            data.append(np.array(leads))
            
        self.data = torch.tensor(np.array(data), dtype=torch.float32)

    def __len__(self):
        return self.num_samples

    def __getitem__(self, idx):
        full_12_lead = self.data[idx]
        input_3_lead = full_12_lead[[1, 6, 10], :] 
        return input_3_lead, full_12_lead

# Instantiate dataset
fs = 500 # Sampling rate 500 Hz
dataset = SyntheticECGDataset(num_samples=10, sampling_rate=fs)
x_batch, y_batch = dataset[0]
target_lead = y_batch[0].numpy()
t = np.linspace(0, len(target_lead)/fs, len(target_lead))
```

```python
# Compute FFT and PSD
freqs = np.fft.fftfreq(len(target_lead), 1/fs)
fft_vals = np.fft.fft(target_lead)
psd = np.abs(fft_vals)**2

# Keep only positive frequencies up to Nyquist (250 Hz)
pos_mask = (freqs > 0) & (freqs < (fs / 2))
freqs_pos = freqs[pos_mask]
psd_pos = psd[pos_mask]

# Interactive Subplot: Time Domain vs Frequency Domain
fig_fft = make_subplots(rows=1, cols=2, subplot_titles=("Time Domain (Lead I)", "Frequency Domain (PSD)"))

fig_fft.add_trace(go.Scatter(x=t, y=target_lead, mode='lines', name='Raw ECG', line=dict(color='#38bdf8')), row=1, col=1)
fig_fft.add_trace(go.Scatter(x=freqs_pos, y=10 * np.log10(psd_pos + 1e-10), mode='lines', name='PSD (dB)', line=dict(color='#f43f5e')), row=1, col=2)

# Highlight typical QRS frequency band
fig_fft.add_vrect(x0=10, x1=25, fillcolor="rgba(16, 185, 129, 0.2)", layer="below", line_width=0, annotation_text="QRS Band", row=1, col=2)

fig_fft.update_layout(
    title="Exploratory Data Analysis: Time vs. Frequency Domain",
    template="plotly_dark",
    xaxis_title="Time (s)",
    xaxis2_title="Frequency (Hz)",
    yaxis2_title="Power (dB)",
    height=400, margin=dict(l=20, r=20, t=60, b=20)
)
fig_fft.show()
```

```python
# Design a 3rd order Butterworth bandpass filter (0.5 - 45 Hz)
nyquist = 0.5 * fs
low = 0.5 / nyquist
high = 45.0 / nyquist
b, a = signal.butter(3, [low, high], btype='band')

# Apply filter using filtfilt for zero phase shift
filtered_lead = signal.filtfilt(b, a, target_lead)

fig_filt = go.Figure()
fig_filt.add_trace(go.Scatter(x=t, y=target_lead, mode='lines', name='Raw (Noisy + Drift)', line=dict(color='#94a3b8', width=1.5), opacity=0.6))
fig_filt.add_trace(go.Scatter(x=t, y=filtered_lead, mode='lines', name='Filtered (Bandpass 0.5-45 Hz)', line=dict(color='#10b981', width=2)))

fig_filt.update_layout(
    title="Effect of Butterworth Bandpass Filtering",
    xaxis_title="Time (s)",
    yaxis_title="Amplitude (mV)",
    template="plotly_dark",
    height=400, margin=dict(l=20, r=20, t=40, b=20)
)
fig_filt.show()
```
