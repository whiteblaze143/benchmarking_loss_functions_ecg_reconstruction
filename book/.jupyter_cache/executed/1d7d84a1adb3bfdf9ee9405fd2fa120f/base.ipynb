```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import plotly.graph_objects as go
import numpy as np

class SpectralSTFTLoss(nn.Module):
    """
    Short-Time Fourier Transform (STFT) Spectral Loss.
    Ensures frequency-domain magnitude preservation across all cardiac harmonics.
    """
    def __init__(self, n_fft=64, hop_length=16):
        super().__init__()
        self.n_fft = n_fft
        self.hop_length = hop_length

    def forward(self, pred, target):
        batch, leads, time = pred.shape
        pred_flat = pred.view(-1, time)
        target_flat = target.view(-1, time)

        pred_stft = torch.stft(pred_flat, n_fft=self.n_fft, hop_length=self.hop_length, return_complex=True)
        target_stft = torch.stft(target_flat, n_fft=self.n_fft, hop_length=self.hop_length, return_complex=True)

        pred_mag = torch.abs(pred_stft)
        target_mag = torch.abs(target_stft)

        return F.l1_loss(pred_mag, target_mag)

class PearsonCorrelationLoss(nn.Module):
    """Pearson Correlation Coefficient Loss across time channels."""
    def __init__(self, eps=1e-8):
        super().__init__()
        self.eps = eps

    def forward(self, pred, target):
        pred_centered = pred - pred.mean(dim=-1, keepdim=True)
        target_centered = target - target.mean(dim=-1, keepdim=True)

        pred_std = torch.sqrt(torch.sum(pred_centered ** 2, dim=-1) + self.eps)
        target_std = torch.sqrt(torch.sum(target_centered ** 2, dim=-1) + self.eps)

        corr = torch.sum(pred_centered * target_centered, dim=-1) / (pred_std * target_std)
        return 1.0 - corr.mean()

class HighFrequencyDerivativeLoss(nn.Module):
    """Penalizes derivative discrepancies to preserve sharp QRS peaks."""
    def forward(self, pred, target):
        pred_diff = pred[:, :, 1:] - pred[:, :, :-1]
        target_diff = target[:, :, 1:] - target[:, :, :-1]
        return F.l1_loss(pred_diff, target_diff)

class KernelMMDLoss(nn.Module):
    """Biased RBF-kernel MMD over flattened record representations."""
    def __init__(self, bandwidths=(0.5, 1.0, 2.0)):
        super().__init__()
        self.bandwidths = bandwidths

    def forward(self, pred, target):
        x = pred.flatten(1)
        y = target.flatten(1)
        d_xx = torch.cdist(x, x).pow(2)
        d_yy = torch.cdist(y, y).pow(2)
        d_xy = torch.cdist(x, y).pow(2)
        mmd = pred.new_zeros(())
        for bandwidth in self.bandwidths:
            scale = 2 * bandwidth ** 2
            mmd = mmd + (
                torch.exp(-d_xx / scale).mean()
                + torch.exp(-d_yy / scale).mean()
                - 2 * torch.exp(-d_xy / scale).mean()
            )
        return mmd / len(self.bandwidths)

class KorsVCGLoss(nn.Module):
    """Transforms 12-lead ECG into 3D Vectorcardiogram (VCG) space using Dower transformation."""
    def __init__(self):
        super().__init__()
        dower_mat = torch.tensor([
            [-0.172, -0.074, 0.122, 0.231, 0.239, 0.194, 0.156, -0.010, -0.089, -0.215, 0.010, 0.000],
            [ 0.057, -0.019, -0.106, -0.022, 0.041, 0.048, -0.227, 0.887, 0.057, -0.061, 0.000, 0.000],
            [-0.229, -0.310, -0.017, -0.021, -0.025, 0.063, 0.022, 0.102, 0.513, 0.387, 0.000, 0.000]
        ], dtype=torch.float32)
        self.register_buffer('dower_mat', dower_mat)

    def forward(self, pred, target):
        batch, leads, time = pred.shape
        pred_flat = pred.transpose(1, 2).reshape(-1, leads)
        target_flat = target.transpose(1, 2).reshape(-1, leads)

        pred_vcg = torch.matmul(pred_flat, self.dower_mat.T)
        target_vcg = torch.matmul(target_flat, self.dower_mat.T)

        return F.mse_loss(pred_vcg, target_vcg)

class CombinatorialCompositeLoss(nn.Module):
    """Wrapper aggregating all 6 loss components."""
    def __init__(self):
        super().__init__()
        self.mse = nn.MSELoss()
        self.corr = PearsonCorrelationLoss()
        self.deriv = HighFrequencyDerivativeLoss()
        self.vcg = KorsVCGLoss()
        self.spectral = SpectralSTFTLoss()
        self.mmd = KernelMMDLoss()

    def forward(self, pred, target):
        l_mse = self.mse(pred, target)
        l_corr = self.corr(pred, target)
        l_deriv = self.deriv(pred, target)
        l_vcg = self.vcg(pred, target)
        l_spectral = self.spectral(pred, target)
        l_mmd = self.mmd(pred, target)

        total_loss = (
            10.0 * l_mse + 2.0 * l_corr + 5.0 * l_deriv
            + 1.0 * l_vcg + 1.0 * l_spectral + 1.0 * l_mmd
        )
        return total_loss, {
            "mse": l_mse.item(), "corr": l_corr.item(), 
            "deriv": l_deriv.item(), "vcg": l_vcg.item(),
            "spectral": l_spectral.item(), "mmd": l_mmd.item()
        }

# Verify Combinatorial Loss execution
loss_fn = CombinatorialCompositeLoss()
pred_dummy = torch.randn(4, 12, 1000, requires_grad=True)
target_dummy = torch.randn(4, 12, 1000)

loss, metrics = loss_fn(pred_dummy, target_dummy)
loss.backward()

print(f"Total Combinatorial Loss: {loss.item():.4f}")
print("Component Breakdown:", metrics)
```
