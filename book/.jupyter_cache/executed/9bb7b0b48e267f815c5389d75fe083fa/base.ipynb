```python
#| label: kde-distribution-mmd
import numpy as np
from scipy import stats
import plotly.graph_objects as go

# Simulate Ground Truth distribution P (bimodal distribution of QRS amplitudes)
np.random.seed(42)
gt_p = np.concatenate([np.random.normal(0.8, 0.15, 500), np.random.normal(1.8, 0.25, 500)])

# 1. MSE-only model prediction (Variance shrinkage: distribution collapsed toward global mean 1.3)
mse_q = np.random.normal(1.3, 0.15, 1000)

# 2. MMD-regularized model prediction (Preserves true bimodal shape of distribution P)
mmd_q = np.concatenate([np.random.normal(0.79, 0.16, 500), np.random.normal(1.78, 0.26, 500)])

# Compute Gaussian Kernel Density Estimates (KDE)
x_eval = np.linspace(0.0, 3.0, 500)
kde_gt = stats.gaussian_kde(gt_p)(x_eval)
kde_mse = stats.gaussian_kde(mse_q)(x_eval)
kde_mmd = stats.gaussian_kde(mmd_q)(x_eval)

# Plot Interactive Plotly Distribution Density Overlay
fig_kde = go.Figure()
fig_kde.add_trace(go.Scatter(x=x_eval, y=kde_gt, mode='lines', line=dict(color='#38bdf8', width=3), name='Ground Truth Distribution P'))
fig_kde.add_trace(go.Scatter(x=x_eval, y=kde_mse, mode='lines', line=dict(color='#f43f5e', width=2, dash='dash'), name='MSE-Only Model Q (Mode Collapse)'))
fig_kde.add_trace(go.Scatter(x=x_eval, y=kde_mmd, mode='lines', line=dict(color='#10b981', width=2), name='MMD Kernel Model Q (RKHS Density Matched)'))

fig_kde.update_layout(
    title="DIDACTIC SIMULATION — KDE View of Stipulated Collapse and Matching",
    xaxis_title="Waveform Amplitude / Interval Feature",
    yaxis_title="Probability Density",
    template="plotly_dark",
    height=450,
    margin=dict(l=20, r=20, t=50, b=20)
)
fig_kde.show()
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import plotly.graph_objects as go
import numpy as np

class PearsonCorrelationLoss(nn.Module):
    """Pearson Correlation Coefficient Loss."""
    def __init__(self, eps=1e-8):
        super().__init__()
        self.eps = eps

    def forward(self, pred, target):
        pred_centered = pred - pred.mean(dim=-1, keepdim=True)
        target_centered = target - target.mean(dim=-1, keepdim=True)

        pred_std = torch.sqrt(torch.sum(pred_centered ** 2, dim=-1) + self.eps)
        target_std = torch.sqrt(torch.sum(target_centered ** 2, dim=-1) + self.eps)

        corr = torch.sum(pred_centered * target_centered, dim=-1) / (pred_std * target_std)
        return 1.0 - corr.mean()

class HighFrequencyDerivativeLoss(nn.Module):
    """First derivative slope penalty."""
    def forward(self, pred, target):
        pred_diff = pred[:, :, 1:] - pred[:, :, :-1]
        target_diff = target[:, :, 1:] - target[:, :, :-1]
        return F.l1_loss(pred_diff, target_diff)

class EinthovenConsistencyLoss(nn.Module):
    """Enforces Lead II = Lead I + Lead III."""
    def forward(self, pred):
        # Index 0: Lead I, Index 1: Lead II, Index 2: Lead III
        lead_I = pred[:, 0, :]
        lead_II = pred[:, 1, :]
        lead_III = pred[:, 2, :]
        
        einthoven_err = lead_II - (lead_I + lead_III)
        return torch.mean(einthoven_err ** 2)

class KorsVCGLoss(nn.Module):
    """Kors Dower 3D VCG Projection Loss."""
    def __init__(self):
        super().__init__()
        # 3x12 Dower Transformation Matrix registered as a non-trainable PyTorch buffer
        dower_mat = torch.tensor([
            [-0.172, -0.074, 0.122, 0.231, 0.239, 0.194, 0.156, -0.010, -0.089, -0.215, 0.010, 0.000],
            [ 0.057, -0.019, -0.106, -0.022, 0.041, 0.048, -0.227, 0.887, 0.057, -0.061, 0.000, 0.000],
            [-0.229, -0.310, -0.017, -0.021, -0.025, 0.063, 0.022, 0.102, 0.513, 0.387, 0.000, 0.000]
        ], dtype=torch.float32)
        self.register_buffer('dower_mat', dower_mat)

    def forward(self, pred, target):
        batch, leads, time = pred.shape
        pred_flat = pred.transpose(1, 2).reshape(-1, leads)
        target_flat = target.transpose(1, 2).reshape(-1, leads)

        # Matrix multiplication to project 12-lead -> 3D VCG (X, Y, Z)
        pred_vcg = torch.matmul(pred_flat, self.dower_mat.T)
        target_vcg = torch.matmul(target_flat, self.dower_mat.T)

        return F.mse_loss(pred_vcg, target_vcg)

class EnergyDifferenceLoss(nn.Module):
    """Regularizes total signal energy norm."""
    def forward(self, pred, target):
        pred_energy = torch.sum(pred ** 2, dim=-1)
        target_energy = torch.sum(target ** 2, dim=-1)
        return F.l1_loss(pred_energy, target_energy)

class CombinatorialCompositeLoss(nn.Module):
    """Master Combinatorial Loss Aggregator."""
    def __init__(self):
        super().__init__()
        self.mse = nn.MSELoss()
        self.corr = PearsonCorrelationLoss()
        self.deriv = HighFrequencyDerivativeLoss()
        self.einthoven = EinthovenConsistencyLoss()
        self.vcg = KorsVCGLoss()
        self.ed = EnergyDifferenceLoss()

    def forward(self, pred, target):
        l_mse = self.mse(pred, target)
        l_corr = self.corr(pred, target)
        l_deriv = self.deriv(pred, target)
        l_einthoven = self.einthoven(pred)
        l_vcg = self.vcg(pred, target)
        l_ed = self.ed(pred, target)

        total_loss = (
            10.0 * l_mse + 
            2.0 * l_corr + 
            5.0 * l_deriv + 
            1.0 * l_einthoven + 
            1.0 * l_vcg + 
            0.5 * l_ed
        )
        return total_loss, {
            "mse": l_mse.item(), "corr": l_corr.item(), 
            "deriv": l_deriv.item(), "einthoven": l_einthoven.item(),
            "vcg": l_vcg.item(), "ed": l_ed.item()
        }

# Verify forward & backward autograd pass
loss_fn = CombinatorialCompositeLoss()
pred_dummy = torch.randn(4, 12, 1000, requires_grad=True)
target_dummy = torch.randn(4, 12, 1000)
loss_val, metrics = loss_fn(pred_dummy, target_dummy)
loss_val.backward()
print("Combinatorial Composite Loss Autograd Gradient Check Passed!")
print(f"Total Loss Value: {loss_val.item():.4f}")

# Plot Gradient Norm Magnitudes per Loss Component
loss_names = list(metrics.keys())
loss_magnitudes = [10.0 * metrics['mse'], 2.0 * metrics['corr'], 5.0 * metrics['deriv'], 
                   1.0 * metrics['einthoven'], 1.0 * metrics['vcg'], 0.5 * metrics['ed']]

fig_grad = go.Figure(data=[go.Bar(
    x=loss_names, y=loss_magnitudes,
    marker_color=['#38bdf8', '#f43f5e', '#10b981', '#fbbf24', '#a855f7', '#ec4899']
)])
fig_grad.update_layout(
    title="Combinatorial Composite Loss Component Gradient Contribution (Weighted Magnitude)",
    xaxis_title="Loss Component",
    yaxis_title="Weighted Loss Magnitude",
    template="plotly_dark",
    height=400, margin=dict(l=20, r=20, t=50, b=20)
)
fig_grad.show()
```
